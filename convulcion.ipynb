CONVULCIONES CON MATRIZ KERNEL

```python
from PIL.ImageFilter import Kernel
import cv2
import matplotlib.pyplot as plt
import numpy as np
img=cv2.imread('PDI2026/Clase1/chanki.jpg', 0)
Kernel=np.array([[0,-1,0],[-1,5,-1], [0,-1,0]]) #mascara de convulcion
img_fin=cv2.filter2D(img,-1,Kernel)
plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.subplot(122)
plt.imshow(img_fin, cmap='gray')
plt.show()
```

CONVOLUCIONES CON MATRIZ DE NITIDEZ A COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('PDI2026/Clase1/chanki.jpg')
# Mascara de Convolucion
kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])

img_fin = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.title('Mascara de Convolucion')

plt.show()
```

TODOS LOS FILTROS DE CONVULCION EN UNO SOLO

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_color = cv2.imread('PDI2026/Clase1/chanki.jpg')
imagen_grises = cv2.imread('PDI2026/Clase1/chanki.jpg', 0)

kernel_enfoque = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]])

kernel_desenfoque = np.ones((7, 7), np.float32)/49

kernel_realce_bordes = np.array([[0, 0, 0],
                                 [-1, 1, 0],
                                 [0, 0, 0]])

kernel_repujado = np.array([[-2, -1, 0],
                            [-1, 1, 1],
                            [0, 1, 2]])

kernel_deteccion_bordes = np.array([[0, 1, 0],
                                    [1, -4, 1],
                                    [0, 1, 0]])

kernel_sobel = np.array([[-1, 0, 1],
                         [-2, 0, 2],
                         [-1, 0, 1]])

kernel_sharpen = np.array([[1, -2, 1],
                           [-2, 5, -2],
                           [1, -2, 1]])

kernel_norte = np.array([[1, 1, 1],
                         [1, -2, 1],
                         [-1, -1, -1]])

kernel_este = np.array([[-1, 1, 1],
                        [-1, -2, 1],
                        [-1, 1, 1]])

kernel_gauss = np.array([[1, 2, 3, 2, 1],
                         [2, 7, 11, 7, 2],
                         [3, 11, 17, 11, 3],
                         [2, 7, 11, 7, 2],
                         [1, 2, 3, 2, 1]], np.float32)/25

filtros = [
    ('Enfoque', kernel_enfoque),
    ('Desenfoque', kernel_desenfoque),
    ('Realce de bordes', kernel_realce_bordes),
    ('Repujado', kernel_repujado),
    ('Deteccion de bordes', kernel_deteccion_bordes),
    ('Filtro de tipo Sobel', kernel_sobel),
    ('Filtro de tipo Sharpen', kernel_sharpen),
    ('Filtro Norte', kernel_norte),
    ('Filtro Este', kernel_este),
    ('Filtro de tipo Gauss', kernel_gauss)
]

# IMPRIMIR LAS ORIGINALES
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(imagen_color, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original a Color')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(imagen_grises, cmap='gray')
plt.title('Imagen Original en Escala Grises')
plt.axis('off')

for nombre_filtro, kernel in filtros:
    imagen_colores_fin = cv2.filter2D(imagen_color, -1, kernel)
    imagen_grises_fin = cv2.filter2D(imagen_grises, -1, kernel)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(imagen_colores_fin, cv2.COLOR_BGR2RGB))
    plt.title(nombre_filtro + ' a COLORES')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(imagen_grises_fin, cmap='gray')
    plt.title(nombre_filtro + ' en ESCALA DE GRISES')
    plt.axis('off')

    plt.show()
```
