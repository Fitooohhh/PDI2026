TODOS LOS FILTROS DE CONVULCION EN UNO SOLO

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_color = cv2.imread('PDI2026/Clase1/chanki.jpg')
imagen_grises = cv2.imread('PDI2026/Clase1/chanki.jpg', 0)

kernel_enfoque = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]])

kernel_desenfoque = np.ones((7, 7), np.float32)/49

kernel_realce_bordes = np.array([[0, 0, 0],
                                 [-1, 1, 0],
                                 [0, 0, 0]])

kernel_repujado = np.array([[-2, -1, 0],
                            [-1, 1, 1],
                            [0, 1, 2]])

kernel_deteccion_bordes = np.array([[0, 1, 0],
                                    [1, -4, 1],
                                    [0, 1, 0]])

kernel_sobel = np.array([[-1, 0, 1],
                         [-2, 0, 2],
                         [-1, 0, 1]])

kernel_sharpen = np.array([[1, -2, 1],
                           [-2, 5, -2],
                           [1, -2, 1]])

kernel_norte = np.array([[1, 1, 1],
                         [1, -2, 1],
                         [-1, -1, -1]])

kernel_este = np.array([[-1, 1, 1],
                        [-1, -2, 1],
                        [-1, 1, 1]])

kernel_gauss = np.array([[1, 2, 3, 2, 1],
                         [2, 7, 11, 7, 2],
                         [3, 11, 17, 11, 3],
                         [2, 7, 11, 7, 2],
                         [1, 2, 3, 2, 1]], np.float32)/25

filtros = [
    ('Enfoque', kernel_enfoque),
    ('Desenfoque', kernel_desenfoque),
    ('Realce de bordes', kernel_realce_bordes),
    ('Repujado', kernel_repujado),
    ('Deteccion de bordes', kernel_deteccion_bordes),
    ('Filtro de tipo Sobel', kernel_sobel),
    ('Filtro de tipo Sharpen', kernel_sharpen),
    ('Filtro Norte', kernel_norte),
    ('Filtro Este', kernel_este),
    ('Filtro de tipo Gauss', kernel_gauss)
]

# IMPRIMIR LAS ORIGINALES
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(imagen_color, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original a Color')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(imagen_grises, cmap='gray')
plt.title('Imagen Original en Escala Grises')
plt.axis('off')

for nombre_filtro, kernel in filtros:
    imagen_colores_fin = cv2.filter2D(imagen_color, -1, kernel)
    imagen_grises_fin = cv2.filter2D(imagen_grises, -1, kernel)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(imagen_colores_fin, cv2.COLOR_BGR2RGB))
    plt.title(nombre_filtro + ' a COLORES')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(imagen_grises_fin, cmap='gray')
    plt.title(nombre_filtro + ' en ESCALA DE GRISES')
    plt.axis('off')

    plt.show()
```

VERTICAL EN GRISES Y COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_v(imagen, tamano_kernel):
    kernel = np.ones((tamano_kernel, 1), np.float32) / tamano_kernel
    imagen_filtrada = cv2.filter2D(imagen, -1, kernel)
    return imagen_filtrada

imagen_bgr = cv2.imread('img1h.png')
imagen_grises = cv2.imread('img1h.png', 0)

tamano_kernel = 55

imagen_vertical_color = convolucion_media_v(imagen_bgr, tamano_kernel)
imagen_vertical_grises = convolucion_media_v(imagen_grises, tamano_kernel)

plt.figure(figsize=(15,10))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB))
plt.title('Original Color')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(imagen_grises, cmap='gray')
plt.title('Original Grises')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(imagen_vertical_color, cv2.COLOR_BGR2RGB))
plt.title('Vertical Color')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(imagen_vertical_grises, cmap='gray')
plt.title('Vertical Grises')
plt.axis('off')

plt.show()
```

DIAGONAL GRISES Y A COLOR

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_d(imagen, tamano_kernel):
    kernel = np.zeros((tamano_kernel, tamano_kernel), np.float32)
    
    for i in range(tamano_kernel):
        kernel[i, i] = 1
    
    kernel = kernel / tamano_kernel
    
    imagen_filtrada = cv2.filter2D(imagen, -1, kernel)
    return imagen_filtrada

imagen_bgr = cv2.imread('img1h.png')
imagen_grises = cv2.imread('img1h.png', 0)

tamano_kernel = 55

imagen_diagonal_color = convolucion_media_d(imagen_bgr, tamano_kernel)
imagen_diagonal_grises = convolucion_media_d(imagen_grises, tamano_kernel)

plt.figure(figsize=(15,10))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB))
plt.title('Original Color')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(imagen_grises, cmap='gray')
plt.title('Original Grises')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(imagen_diagonal_color, cv2.COLOR_BGR2RGB))
plt.title('Diagonal Color')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(imagen_diagonal_grises, cmap='gray')
plt.title('Diagonal Grises')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_d_inversa(imagen, tamano_kernel):
    kernel = np.zeros((tamano_kernel, tamano_kernel), np.float32)
    
    for i in range(tamano_kernel):
        kernel[i, tamano_kernel - i - 1] = 1
    
    kernel = kernel / tamano_kernel
    
    imagen_filtrada = cv2.filter2D(imagen, -1, kernel)
    return imagen_filtrada

imagen_bgr = cv2.imread('img1h.png')
imagen_grises = cv2.imread('img1h.png', 0)

tamano_kernel = 55

imagen_diagonal_inv_color = convolucion_media_d_inversa(imagen_bgr, tamano_kernel)
imagen_diagonal_inv_grises = convolucion_media_d_inversa(imagen_grises, tamano_kernel)

plt.figure(figsize=(15,10))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB))
plt.title('Original Color')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(imagen_grises, cmap='gray')
plt.title('Original Grises')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(imagen_diagonal_inv_color, cv2.COLOR_BGR2RGB))
plt.title('Diagonal Inversa Color')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(imagen_diagonal_inv_grises, cmap='gray')
plt.title('Diagonal Inversa Grises')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_bgr = cv2.imread('img1h.png')
imagen_grises = cv2.imread('img1h.png', 0)

def convolucion_media_h(imagen, tamano_kernel):
    kernel = np.ones((1, tamano_kernel), np.float32) / tamano_kernel
    return cv2.filter2D(imagen, -1, kernel)

def convolucion_media_v(imagen, tamano_kernel):
    kernel = np.ones((tamano_kernel, 1), np.float32) / tamano_kernel
    return cv2.filter2D(imagen, -1, kernel)

def convolucion_media_d(imagen, tamano_kernel):
    kernel = np.zeros((tamano_kernel, tamano_kernel), np.float32)
    for i in range(tamano_kernel):
        kernel[i, i] = 1
    kernel = kernel / tamano_kernel
    return cv2.filter2D(imagen, -1, kernel)

def convolucion_media_d_inversa(imagen, tamano_kernel):
    kernel = np.zeros((tamano_kernel, tamano_kernel), np.float32)
    for i in range(tamano_kernel):
        kernel[i, tamano_kernel - i - 1] = 1
    kernel = kernel / tamano_kernel
    return cv2.filter2D(imagen, -1, kernel)

def imprimir_fila(imagen_color, imagen_gris, titulo_color, titulo_gris):
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(imagen_color, cv2.COLOR_BGR2RGB))
    plt.title(titulo_color)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(imagen_gris, cmap='gray')
    plt.title(titulo_gris)
    plt.axis('off')

    plt.tight_layout()
    plt.show()

tamano_kernel = 55

imagen_horizontal_color = convolucion_media_h(imagen_bgr, tamano_kernel)
imagen_horizontal_grises = convolucion_media_h(imagen_grises, tamano_kernel)

imagen_vertical_color = convolucion_media_v(imagen_bgr, tamano_kernel)
imagen_vertical_grises = convolucion_media_v(imagen_grises, tamano_kernel)

imagen_diagonal_color = convolucion_media_d(imagen_bgr, tamano_kernel)
imagen_diagonal_grises = convolucion_media_d(imagen_grises, tamano_kernel)

imagen_diagonal_inv_color = convolucion_media_d_inversa(imagen_bgr, tamano_kernel)
imagen_diagonal_inv_grises = convolucion_media_d_inversa(imagen_grises, tamano_kernel)

imprimir_fila(imagen_bgr, imagen_grises, 'Original Color', 'Original Gris')
imprimir_fila(imagen_horizontal_color, imagen_horizontal_grises, 'Horizontal Color', 'Horizontal Gris')
imprimir_fila(imagen_vertical_color, imagen_vertical_grises, 'Vertical Color', 'Vertical Gris')
imprimir_fila(imagen_diagonal_color, imagen_diagonal_grises, 'Diagonal Color', 'Diagonal Gris')
imprimir_fila(imagen_diagonal_inv_color, imagen_diagonal_inv_grises, 'Diagonal Inversa Color', 'Diagonal Inversa Gris')
```
