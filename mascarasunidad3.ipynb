MASCARAS EN ESCALA DE GRISES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('southpark.jpg', 0)
alto, ancho = imagen.shape[:2]
mascara = np.zeros_like(imagen)
centro = (ancho//2, alto//2)
radio = 300
cv2.circle(mascara, centro, radio, (255), -1)
img_mascara = cv2.bitwise_and(imagen, imagen, mask=mascara)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(imagen, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(img_mascara, cmap='gray')
plt.title('Mascara en Grises')

plt.show()
```

MÁSCARA A COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('southpark.jpg')
alto, ancho = imagen.shape[:2]

mascara = np.zeros((alto, ancho), dtype=np.uint8)
centro = (ancho//2, alto//2)
radio = 300
cv2.circle(mascara, centro, radio, (255), -1)

img_mascara = cv2.bitwise_and(imagen, imagen, mask=mascara)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_mascara, cv2.COLOR_BGR2RGB))
plt.title('Mascara en Colores')

plt.show()
```
