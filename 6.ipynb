Desarrolla una aplicación que permita identificar si una palabra es un anagrama de otra
(Bolo -Lobo)

```python
# Ejercicio 6: Identificar si dos palabras son anagramas

print("Verificación de anagramas")
palabra1 = input("Introduce la primera palabra: ")
palabra2 = input("Introduce la segunda palabra: ")

# Convertir a minúsculas
palabra1 = palabra1.lower()
palabra2 = palabra2.lower()

# Verificar si tienen la misma longitud
if len(palabra1) != len(palabra2):
    print(f"'{palabra1}' y '{palabra2}' NO son anagramas")
else:
    # Contar letras de la primera palabra
    letras_palabra1 = {}
    for letra in palabra1:
        if letra in letras_palabra1:
            letras_palabra1[letra] += 1
        else:
            letras_palabra1[letra] = 1
    
    # Contar letras de la segunda palabra
    letras_palabra2 = {}
    for letra in palabra2:
        if letra in letras_palabra2:
            letras_palabra2[letra] += 1
        else:
            letras_palabra2[letra] = 1
    
    # Comparar si tienen las mismas letras con la misma frecuencia
    if letras_palabra1 == letras_palabra2:
        print(f"'{palabra1}' y '{palabra2}' SON anagramas")
    else:
        print(f"'{palabra1}' y '{palabra2}' NO son anagramas")
```
