Foto 1, aqui se puede apreciar al querido amigo Pérez siendo Pérez

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np


img_bgr = cv2.imread('Hardy.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gris = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)


cv2.imwrite('Hardy_gris.jpg', img_gris)
cv2.imwrite('Hardy_bgr.jpg', img_bgr)

print("✓ Hardy.jpg cargado y procesado")


plt.figure(figsize=(20, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_rgb)
plt.title('Original (RGB)')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(img_gris, cmap='gray')
plt.title('Escala de Grises')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(img_rgb[..., 0], cmap='gray')
plt.title('Canal Rojo (Gris)')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(img_rgb[..., 0], cmap='Reds')
plt.title('Canal Rojo (Color Falso)')
plt.axis('off')

plt.suptitle('Análisis de Hardy - Modo 1', fontsize=16)
plt.show()


plt.figure(figsize=(20, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_rgb[..., 1], cmap='gray')
plt.title('Canal Verde (Gris)')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(img_rgb[..., 1], cmap='Greens')
plt.title('Canal Verde (Color Falso)')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(img_rgb[..., 2], cmap='gray')
plt.title('Canal Azul (Gris)')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(img_rgb[..., 2], cmap='Blues')
plt.title('Canal Azul (Color Falso)')
plt.axis('off')

plt.suptitle('Análisis de presi - Modo 2', fontsize=16)
plt.show()

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('imagen original')
plt.imshow(img_rgb)

plt.subplot(1, 2, 2)
plt.title('Hardy - Escala de Grises')
plt.imshow(img_gris, cmap='gray')

plt.show()
```
