ploteo por canales

```python
import cv2
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=[10,5]
img=cv2.imread('chanki.jpg', cv2.IMREAD_COLOR)
#IMAGEN ORIGINAL Y DIVIDIVA POR CANALES

plt.subplot(141)
plt.title('imagen original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title('CANAL ROJO')
plt.imshow(img[:,:,0],cmap='gray')

plt.subplot(143)
plt.title('CANAL VERDE')
plt.imshow(img[:,:,1],cmap='gray')

plt.subplot(144)
plt.title('CANAL AZUL')
plt.imshow(img[:,:,2],cmap='Blues')

plt.show()
```

con colores relleno

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize']=[10,5]
img=cv2.imread('chanki.jpg', cv2.IMREAD_COLOR)
imgRGB=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

r=img[:,:,2]
g=img[:,:,1]
b=img[:,:,0]

#crear imagen por canal
r_img=np.zeros_like(imgRGB)
r_img[:,:,0]=r

g_img=np.zeros_like(imgRGB)
g_img[:,:,1]=g

b_img=np.zeros_like(imgRGB)
b_img[:,:,2]=b

plt.figure(figsize=(10,5))

plt.subplot(141)
plt.title('imagen original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('imagen roja')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('imagen verde')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title('imagen azul')
plt.imshow(b_img)
plt.axis('off')

plt.show()
```

Generecion de falso color con Cmap

```python
import cv2
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=[10,5]
img=cv2.imread('chanki.jpg', cv2.IMREAD_COLOR)

plt.figure(figsize=(15,5))

# Primera fila
plt.subplot(2,6,1)
plt.title('ROJO')
plt.imshow(img[:,:,2],cmap='Reds')
plt.axis('off')

plt.subplot(2,6,2)
plt.title('VERDE')
plt.imshow(img[:,:,1],cmap='Greens')
plt.axis('off')

plt.subplot(2,6,3)
plt.title('AZUL')
plt.imshow(img[:,:,0],cmap='Blues')
plt.axis('off')

plt.subplot(2,6,4)
plt.title('CAFE')
plt.imshow(img[:,:,1],cmap='YlOrBr')
plt.axis('off')

plt.subplot(2,6,5)
plt.title('ROSA')
plt.imshow(img[:,:,2],cmap='RdPu')
plt.axis('off')

plt.subplot(2,6,6)
plt.title('AMARILLO')
plt.imshow(img[:,:,1],cmap='Wistia')
plt.axis('off')

# Segunda fila
plt.subplot(2,6,7)
plt.title('NEGRO')
plt.imshow(img[:,:,0],cmap='gray')
plt.axis('off')

plt.subplot(2,6,8)
plt.title('BLANCO')
plt.imshow(img[:,:,0],cmap='binary')
plt.axis('off')

plt.subplot(2,6,9)
plt.title('MORADO')
plt.imshow(img[:,:,2],cmap='Purples')
plt.axis('off')

plt.subplot(2,6,10)
plt.title('GRIS')
plt.imshow(img[:,:,1],cmap='Greys')
plt.axis('off')

plt.subplot(2,6,11)
plt.title('NARANJA')
plt.imshow(img[:,:,2],cmap='Oranges')
plt.axis('off')

plt.subplot(2,6,12)
plt.title('CYAN')
plt.imshow(img[:,:,0],cmap='GnBu')
plt.axis('off')


plt.tight_layout()
plt.show()
```
