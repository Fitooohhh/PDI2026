Negativo de una imagen

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg', cv2.IMREAD_GRAYSCALE)
imagen_negativo = 255 - imagen # imagen negativa

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(imagen, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(imagen_negativo, cmap='gray')
plt.title('Negativo Grises')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg')
imagen_negativo = 255 - imagen # imagen negativa

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imagen_negativo, cv2.COLOR_BGR2RGB))
plt.title('Negativo Grises')

plt.show()
```

CONTRASTE IMAGEN EN ESCALA DE GRISES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# recibimos imagen en escala de grises
def contraste(img, factor=1.0):
    img = img.astype(np.float32)
    media = np.mean(img)
    img_contraste = media + (img-media)*factor
    img_contraste = np.clip(img_contraste, 0, 255).astype(np.uint8)
    return img_contraste

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg', 0)
contrastesito = 7
imagen_contraste = contraste(imagen, contrastesito)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(imagen, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(imagen_contraste, cmap='gray')
plt.title('Contraste Grises')

plt.show()
```

CONTRASTE IMAGEN A COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# recibimos imagen en BGR
def contraste(img, factor=1.0):
    img = img.astype(np.float32)
    for i in range(3):
        media = np.mean(img[:,:,i])
        img[:,:,i] = media + (img[:,:,i]- media) * factor
        img[:,:,i] = np.clip(img[:,:,i], 0, 255)
    return img.astype(np.uint8)

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg')
contrastesito = 3
imagen_contraste = contraste(imagen, contrastesito)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imagen_contraste, cv2.COLOR_BGR2RGB))
plt.title('Contraste Colores')

plt.show()
```

BRILLO EN UNA IMAGEN EN ESCALA DE GRISES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def brillo(img, brillo=0):
    img_brillo = img.astype(np.float32) + brillo
    img_brillo = np.clip(img_brillo, 0, 255).astype(np.uint8)
    return img_brillo

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg', 0)
brillito = -70 # puede ser valores - o +
imagen_brillo = brillo(imagen, brillito)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(imagen, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(imagen_brillo, cmap='gray')
plt.title('Brillo en Grises')

plt.show()
```

BRILLO EN IMAGEN A COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def brillo_color(img, brillo=0):
    img = img.astype(np.float32)
    for i in range(3):
        img[:,:,i] = img[:,:,i] + brillo
        img[:,:,i] = np.clip(img[:,:,i], 0, 255)
    
    return img.astype(np.uint8)

imagen = cv2.imread('PDI2026/Clase1/chanki.jpg')
brillito = -70
imagen_brillo = brillo_color(imagen, brillito)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imagen_brillo, cv2.COLOR_BGR2RGB))
plt.title('Brillo a Colores')

plt.show()
```
